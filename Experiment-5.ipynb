```python
#Lambda Functions
func = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(func(5))
print(func(-3))
print(func(0))
```

```python
#Using with List Comprehension
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())

#Using for Returning Multiple Results
calc = lambda x, y: (x + y, x * y)
res = calc(3, 4)
print(res)
```

```python
#Sets
s = {1, 2, 3}; s = set([1, 2, 3]); s = set()
s = {10, 50, 20}
print(s)
print(type(s))

# typecasting list to set
s = set(["a", "b", "c"])
print(s)

# a set cannot have duplicate values
s = {"Geeks", "for", "Geeks"}
print(s)
```

```python
#Methods for Sets
s = {"a", "b", "c"}
s.add("d")
s.remove("c")
print(s)

#Union of Sets
a = {"x", "y"}
b = {"y", "z"}
u = a.union(b)
print(a|b)
print(u)

#Intersection of Sets
a = {1, 2, 3}
b = {2, 3, 4}
i = a.intersection(b)
print(a&b)
print(i)

#Difference of Sets
a = {1, 2, 3}
b = {2, 3, 4}
d = a.difference(b)
print(a-b)
print(d)
```

```python
#Symmetric Difference
a = {1, 2, 3}
b = {2, 3, 4}
print(a^b)

#Clearing a Set
s = {1, 2, 3}
s.clear()
print(s)
```

```python
#A is subset of B if all elements of A are in B.
A = {1, 2}
B = {1, 2, 3, 4}
print(A.issubset(B))

#B is superset of A
print(B.issuperset(A))
print(3 in {1, 2, 3})
for item in {1, 2, 3}:
    print(item)
s = {x*x for x in range(5)}
print(s)
```
