```python
# Prime Number checker
num = int(input("Enter a Number :"))
if num <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print(f"{num} is Prime Number")
```

```python
# Even–Odd Check Using Bitwise Operator

num = int(input("Enter a number: "))

if num & 1 == 0:  # binary and is used
    print("Even Number")
else:
    print("Odd Number")
```
