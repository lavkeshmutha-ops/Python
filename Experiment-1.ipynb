```python
# Prime Number checker
 num = int(input("Enter a Number :"))
 if num <= 1:
    print("Not a Prime Number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a Prime Number")
            break
    else:
        print("%d is Prime Number", num)
```
