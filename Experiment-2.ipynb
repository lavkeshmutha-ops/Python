```python

# --- Conditional Statements Demonstration ---

print("--- 1. if statement ---")
# The 'if' statement executes a block of code only if the condition is True.


number = int(input ("Enter a number :"))

if number > 5:
    print(f"The number {number} is greater than 5.")
print("-" * 20 + "\n")
```

```python
print("--- 2. if..else statement ---")
# The 'if..else' statement executes the 'if' block if the condition is True,
# otherwise it executes the 'else' block.
age = 17
if age >= 18:
    print("You are old enough to vote.")
else:
    print("You are not old enough to vote yet.")
print("-" * 20 + "\n")
```

```python
print ("--3. if.. elif..else statement, nested if statement ---")
# The 'if..else' statement executes the 'if' block if the condition is True,
# otherwise it executes the 'else' block.
score = 85;
if score >= 90:
  grade = 'A'
elif score >=  80:
  grade = 'B'
elif score >= 70:
  grade = 'C'
else :
    # The 'if..else' statement executes the 'if' block if the condition is True,
  # otherwise it executes the 'else' block.
  if score >= 60:
      grade = 'D'
  else:
          grade = 'F'
          print ("Your Score of %d results in a grade of %d."%(score, grade ))
```

```python

# --- Looping Statements Demonstration ---

print("--- 4. while loop ---")
# The 'while' loop continues to execute as long as the given condition is True.
count = 0
print("Counting up to 3:")
while count < 4:
    print(count)
    count += 1 # Increment the counter to avoid an infinite loop
print("-" * 20 + "\n")
```

```python
print("--- 5. for loop, nested loop ---")
# The 'for' loop is used to iterate over a sequence (like a list, tuple, string, or range).

# Example 1: Iterating over a range
print("Iterating through numbers 0 to 2 using a for loop:")
for i in range(3):
    print(i)
   #  Example 2: Nested loop
print("\nDemonstrating a nested loop (multiplication table for 1 and 2):")
for multiplier in range(1, 3):
    for number in range(1, 3):
        product = multiplier * number
        print(f"{multiplier} * {number} = {product}")
    print("-" * 10) # Separator for each multiplier loop iteration
print("-" * 20 + "\n")
```
