```python
# Parameters:

#Get the items from a list starting at position 1 and ending at position 4 (exclusive).

a = [1, 2, 3, 4, 5, 6, 7, 8, 9]
# Get elements from index 1 to 4 (excluded)
print(a[1:4])

#Get all the items from a list

a = [1, 2, 3, 4, 5, 6, 7, 8, 9]
# Get all elements in the list
print(a[::])
print(a[:])

#Get all items before/after a specific position
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Get elements starting from index 2
# to the end of the list
b = a[2:]
print(b)

# Get elements starting from index 0
# to index 3 (excluding 3th index)
c = a[:3]
print(c)

#Get items at specified intervals USE STEP PARAMETER
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Get every second element from the list
# starting from the beginning
b = a[::2]
print(b)

# Get every third element from the list
# starting from index 1 to 8(exclusive)
c = a[1:8:3]
print(c)
```

```python
# Negative Indexing
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Get elements starting from index -2
# to end of list
b = a[-2:]
print(b)

# Get elements starting from index 0
# to index -3 (excluding 3th last index)
c = a[:-3]
print(c)

# Get elements from index -4
# to -1 (excluding index -1)
d = a[-4:-1]
print(d)

# Get every 2nd elements from index -8
# to -1 (excluding index -1)
e = a[-8:-1:2]
print(e)
```

```python
# 1️⃣ Creating Lists
print ("---- 1. Creating Lists ----")
# Empty list
empty_list = []
# List of integers
numbers = [10, 20, 30, 40, 50]
# List of String
fruits = ['apple', 'banana', 'cherry']
# Mixed List
mixed = [10, "hello", 3.14, True, [1, 2, 3]]
print("Empty List:", empty_list)
print("Numbers List:", numbers)
print("Fruits List:", fruits)
print("Mixed List:", mixed)
print()

```

```python
# 2️⃣ Basic List Operations
print("----- 2. Basic List Operations -----")
# Concatenation
combined = numbers + [60, 70]
print("Concatenated List:", combined)
# Repetition
repeated = fruits * 2
print("Repeated List:", repeated)
# Membership
print("Is 'banana' in fruits?", "banana" in fruits)
print("Is 'grape' not in fruits?", "grape" not in fruits)
# Length
print("Length of numbers list:", len(numbers))
print()


```

```python
# 3️⃣ Indexing and Slicing
print("----- 3. Indexing and Slicing -----")
colors = ["red", "green", "blue", "yellow", "purple"]
# Indexing
print("First element:", colors[0])
print("Third element:", colors[2])
print("Last element:", colors[-1])
# Slicing
print("Colors[0:3]:", colors[0:3])
print("Colors[2:]:", colors[2:])
print("Colors[:4]:", colors[:4])
print("Colors[-3:]:", colors[-3:])
print()
```

```python
# 4️⃣ Built-in Functions on Lists
print("----- 4. Built-in Functions on Lists -----")
numbers = [15, 8, 22, 5, 13, 30]
print("Numbers:", numbers)
print("Length:", len(numbers))
print("Maximum value:", max(numbers))
print("Minimum value:", min(numbers))
print("Sum of numbers:", sum(numbers))
print("Sorted (ascending):", sorted(numbers))
print("Sorted (descending):", sorted(numbers, reverse=True))
print()
```

```python
# 5️⃣ List Methods
print("----- 5. List Methods -----")
fruits = ["apple", "banana", "cherry"]
print("Original list:", fruits)
# append()
fruits.append("mango")
print("After append:", fruits)

# insert()
fruits.insert(1, "orange")
print("After insert:", fruits)
# remove()
fruits.remove("banana")
print("After remove:", fruits)
# pop()
popped_item = fruits.pop()
print("Popped item:", popped_item)
print("After pop:", fruits)
# index()
print("Index of 'cherry':", fruits.index("cherry"))
# count()
fruits.append("apple")
print("Count of 'apple':", fruits.count("apple"))
# sort()
fruits.sort()
print("After sort:", fruits)
# reverse()
fruits.reverse()
print("After reverse:", fruits)
# copy()
copy_list = fruits.copy()
print("Copied list:", copy_list)
# extend()
more_fruits = ["papaya", "grapes"]
fruits.extend(more_fruits)
print("After extend:", fruits)
# clear()
copy_list.clear()
print("After clear (empty copy):", copy_list)
print()
print("All List Concepts Covered Successfully!")
```
