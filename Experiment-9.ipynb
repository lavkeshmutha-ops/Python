```python
import pandas as pd
import matplotlib.pyplot as plt

# Create dataset using dictionary
data = {
"Name": ["Aanya", "Riya", "Kabir", "Arjun"],
"Maths": [85, 90, 78, 88],
"Science": [92, 87, 80, 85],
"English": [88, 91, 75, 82]
}

# Create DataFrame
df = pd.DataFrame(data)
# Display full DataFrame
print("Full DataFrame:")
print(df)
# Display selected columns
print("\nSelected Columns (Name and Maths):")
print(df[["Name", "Maths"]])
# Plot graph
df.plot(x="Name", y="Maths", kind="bar")
plt.title("Maths Marks of Students")
plt.show()
```

```python
data = {
"Name": ["Amit", "Sneha", "Ravi", "Priya"],
"Age": [20, 21, 19, 22],
"Marks": [85, 90, 78, 88]
}
df = pd.DataFrame(data)
print(df)
```

```python
import matplotlib.pyplot as plt
# Plot Marks

plt.plot(df["Name"], df["Marks"])
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
#Using List of Lists
import pandas as pd
data = [
["Amit", 20, 85],
["Sneha", 21, 90],
["Ravi", 19, 78],
["Priya", 22, 88]
]

df = pd.DataFrame(data, columns=["Name", "Age", "Marks"])
print(df)
print("\n")

#Using List of Dictionaries
data = [
{"Name": "Amit", "Age": 20, "Marks": 85},
{"Name": "Sneha", "Age": 21, "Marks": 90}
]
df = pd.DataFrame(data)
print(df)
```
