```python
# @title Input for Factorial Calculation
user_x = 5 # @param {type:"integer"}
```

You can change the value of `user_x` in the above cell to calculate the factorial of a different number.

```python
def fibonacci(n):
  if n <= 0:
    return []
  elif n == 1:
    return [0]
  else:
    list_fib = [0, 1]
    while len(list_fib) < n:
      next_fib = list_fib[-1] + list_fib[-2]
      list_fib.append(next_fib)
    return list_fib

print(f"Fibonacci sequence up to {user_x} terms: {fibonacci(user_x)}")
```

```python
def fact(x) :
  if x == 0 or x == 1 :
     return 1
  else :
    return x * fact(x-1)

# Example usage to demonstrate output
print(f"Factorial of {user_x} is: {fact(user_x)}")
print(f"Factorial of 0 is: {fact(0)}")
```
