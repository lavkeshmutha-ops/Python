```python
%%writefile my_package/module1.py
def check_prime(num):
    if num == 1:
        print(" prime no")
        return
    if num == 2 or num == 3:
        print("prime no")
        return
    if num % 2 == 0 or num % 3 == 0:
        print("Not prime")
        return

    count = 5 # This assignment is immediately overwritten by the loop variable `count`.
    for count in range(5, num, count + 6):
        if (num % count) == 0 or ((num) % (count + 2) == 0): # Added parentheses for clarity in num % (count+2)
            print("Not prime")
            return
    print("prime no")

def calculate_factorial(num):
    fact = 1
    if num > 0:
        for count in range(1, num + 1):
            fact *= count
        print("\nFactorial of ", num, " is", fact)
    elif num == 0:
        print("\nFactorial of 0 is 1")
    else:
        print("\nPlease enter a non-negative integer !")
```

```python
%%writefile my_package/module2.py
def add(a, b):
    result = a + b
    return result
def sub(a, b):
    result = a - b
    return result
def mul(a, b):
    result = a * b
    return result
def div(a, b):
    result = a / b
    return result
```

```python
from my_package import module1 as m
number = int(input("\nEnter number = "))
print(m.check_prime(number))
print(m.calculate_factorial(number))
```
